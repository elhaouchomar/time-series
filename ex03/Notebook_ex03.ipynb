```python
import pandas as pd
import numpy as np

business_dates = pd.bdate_range('2021-01-01', '2021-12-31')
tickers = ['AAPL', 'FB', 'GE', 'AMZN', 'DAI']
# print(business_dates)
index = pd.MultiIndex.from_product([business_dates, tickers], names=['Date', 'Ticker'])

market_data = pd.DataFrame(index=index,
                          data=np.random.randn(len(index), 1),
                          columns=['Price'])

print(market_data)

df_pivot = market_data.pivot_table(index="Date", columns="Ticker", values="Price", aggfunc='mean')

returns = df_pivot.pct_change()

print(returns.head())
```
