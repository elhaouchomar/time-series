```python
import pandas as pd

df = pd.read_csv("./data/AAPL.csv")

print(df.head())

df["Date"] = pd.to_datetime(df["Date"])

df.set_index("Date", inplace=True)

print(df.isna().sum())
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=[go.Candlestick(
    x=df.index,
    open=df["Open"],
    high=df["High"],
    low=df["Low"],
    close=df["Close"]
)])

fig.show()
```

```python
fig = go.Figure(data=go.Scatter(x=[1,2,3], y=[4,5,6]))
fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=[go.Candlestick(
    x=df.index,
    open=df["Open"],
    high=df["High"],
    low=df["Low"],
    close=df["Close"]
)])
fig.show()
```

```python
monthly = df.resample("BM").agg({
    "Open": "first",
    "High": "max",
    "Low": "min",
    "Close": "last",
    "Volume": "sum"
})

print(monthly)
print("Number of months:", len(monthly))
```

```python
df["return"] = df["Open"].pct_change()
#or
df["return"] = (df["Open"] - df["Open"].shift(1)) / df["Open"].shift(1)
```
